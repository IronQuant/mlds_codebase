### Label transfer

Can other central banks' labelled sentences substitute for the FOMC's own?
One seed (78516), lowercased room; cased runs shown as open markers for
context. Values hardcoded from results.csv -- update after the 3-seed sweep.

Writes `label_transfer.png`.

```python
import matplotlib.pyplot as plt

# seed 78516; both runs lowercased, identical recipe
NL = chr(10)
labels = [f"Fine-tuned on{NL}24 other banks (WCB)", f"Fine-tuned on{NL}FOMC sentences"]
values = [0.6729, 0.7086]

fig, ax = plt.subplots(figsize=(4.2, 3.0))
bars = ax.bar(
    labels,
    values,
    width=0.55,
    color=["white", "0.35"],
    edgecolor="black",
    linewidth=1.0,
)
for b, v in zip(bars, values):
    ax.text(
        b.get_x() + b.get_width() / 2, v + 0.01, f"{v:.3f}", ha="center", fontsize=9
    )

ax.set_ylabel("Macro-F1", fontsize=9)
ax.set_ylim(0, 0.78)
ax.tick_params(axis="x", labelsize=9, length=0)
for sp in ("top", "right"):
    ax.spines[sp].set_visible(False)

fig.tight_layout()
fig.savefig("label_transfer.png", dpi=200, bbox_inches="tight", facecolor="white")
```

### What it shows

Training on 24 other central banks' labels -- zero FOMC labels -- reaches
0.673 vs 0.709 with the FOMC's own 1,984 labels; both lowercased, identical
recipe, seed 78516. The near-equal bars are the point: stance labels
transfer across institutions. Caption carries: single seed, no exact-match
train/test overlap.
